```python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    # This is a placeholder tool -- it doesn't call a real weather API,
    # it just returns a fixed string so we can prove the whole pipeline works.
    return f"It's always sunny in {city}!"

agent = create_agent(
    model="gpt-3.5-turbo",   # Updated to a valid OpenAI model name
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "What's the weather in San Francisco?"}]}
)

# .content_blocks is the standardized way LangChain exposes a message's content --
# we explain exactly what this is and why it exists later in this notebook.
print(result["messages"][-1].content_blocks)
```

The `response` object is a LangChain `AIMessage` object. When you use `pprint` on it, you're seeing a pretty-printed representation of the Python object itself, including its class name and attributes.

If you want to view the content of this object as a JSON string, you typically need to convert it to a dictionary first (if the object has a method for that, like `dict()`), and then use Python's `json` module to format it as a pretty-printed JSON string.

```python
import json

# LangChain message objects are Pydantic models and can be directly converted to a dictionary.
# Use `.dict()` for Pydantic v1, or `.model_dump()` for Pydantic v2.
# Given current LangChain versions, `.dict()` should work, but `.model_dump()` is forward-compatible.
response_dict = response.model_dump() # Or response.dict() if on an older Pydantic version

# print(response_dict)
print(json.dumps(response_dict, indent=2))
```

# Streaming

```python
open_ai_model = init_chat_model("openai:gpt-5-mini")
```

```python
chunks = []
full_message = None
```

```python
import time
for chunk in open_ai_model.stream("Write a long sentence about how AI Agents work"):
  chunks.append(chunk)
  print(repr(chunk.text),type(chunk))
  full_message = chunk if full_message is None else full_message + chunk
  time.sleep(1)

print()
print("Reconstructed full message:", full_message.content)
```

```python
def get_weather(location:str) -> str:
   "Get the weather at a location"
   print("I was called from somewhere")


model_with_tools = open_ai_model.bind_tools([get_weather,set_password])

response = model_with_tools. invoke('set password to admin123' )
```

```python
type(response)
```

```python
for tool_call in response.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f"Args: {tool_call['args']}")
    print(f"ID: {tool_call['id']}")
```

```python
ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "Delhi"},
        "id": "call_123"
    }]
)

weather_result = "Sunny 35 degrees Celsius" # We will call this tool mannualy


tool_result = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"
)


messages = [
    HumanMessage('What is the weather in Delhi?'),
    ai_message,
    tool_result
]

final_response = open_ai_model.invoke(messages)
final_response
print(final_response.content)
```
